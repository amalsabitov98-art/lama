# 🎙️ Клонирование голоса через Replicate (дёшево, надёжно)

Модель XTTS крутится **на серверах Replicate** — у тебя ничего тяжёлого не ставится,
поэтому никаких падений, как было раньше. Платишь копейки за каждую генерацию
(~0.1–0.3 цента), подписки нет.

## Что нужно один раз:
1. Аккаунт на **replicate.com** (можно через GitHub).
2. **Billing** → положить небольшой баланс (хватит и $5 надолго).
3. **Account → API tokens** → скопировать токен.

## Как пользоваться:
Запускай ячейки по порядку (▶️): вставишь токен → загрузишь сэмпл своего голоса
(6–30 сек чистой речи) → впишешь текст → получишь аудио своим голосом.

> 🔐 Токен вводится скрыто и никуда, кроме Replicate, не уходит.

### Шаг 2. Введи свой Replicate API-токен
Ввод скрытый — токен не отобразится.

```python
import os, getpass
os.environ['REPLICATE_API_TOKEN'] = getpass.getpass('Вставь свой Replicate API-токен и нажми Enter: ')
print('✅ Токен принят')
```

### Шаг 3. Загрузи сэмпл своего голоса
6–30 секунд чистой речи, без музыки и шума (wav/mp3/m4a/ogg).
Сэмпл можно достать из своего ролика инструментом-экстрактором.

```python
from google.colab import files
print('Выбери аудиофайл со своим голосом:')
_up = files.upload()
sample_path = list(_up.keys())[0]
print('✅ Сэмпл загружен:', sample_path)
```

### Шаг 4. Впиши текст и получи озвучку 🎧
Меняй текст в поле и запускай ячейку сколько угодно раз.

```python
#@title Текст для озвучки
text = "Привет! Это тест моего голоса. Если ты это слышишь моим голосом — всё получилось." #@param {type:"string"}
language = "ru" #@param ["ru","en","es","fr","de","it","pt","pl","tr","nl","cs","ar","zh-cn","ja","hu","ko","hi"]

import replicate, urllib.request
from IPython.display import Audio, display

print('Генерирую озвучку… (несколько секунд)')
output = replicate.run(
    'lucataco/xtts-v2',
    input={
        'text': text,
        'speaker': open(sample_path, 'rb'),
        'language': language,
    },
)

# output может быть объектом-файлом или URL — обрабатываем оба случая
out_path = 'voice_output.wav'
if hasattr(output, 'read'):
    data = output.read()
else:
    data = urllib.request.urlopen(str(output)).read()
with open(out_path, 'wb') as f:
    f.write(data)

print('✅ Готово! Слушай ниже, файл сохранён как', out_path)
display(Audio(out_path))
try:
    files.download(out_path)  # авто-скачивание
except Exception:
    pass
```
